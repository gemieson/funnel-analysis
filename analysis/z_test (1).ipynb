## Test 1: Organic vs Referral

```python
from statsmodels.stats.proportion import proportions_ztest

counts  = [1698, 1202]
nobs    = [3111, 1863]
z, p = proportions_ztest(counts, nobs)
print(f"Channel gap:  z = {z:.2f}, p = {p:.2e}")
```

## Test 2: Return Rates of Droppers VS Completers

```python
counts  = [round(0.097 * 3965), round(0.404 * 5749)]
nobs    = [3965, 5749]
z, p = proportions_ztest(counts, nobs)
print(f"Return rates: z = {z:.2f}, p = {p:.2e}")
```
