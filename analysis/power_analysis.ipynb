```python
# analysis/power_analysis.py
# Power analysis for the cart-recovery A/B test.

import math
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

BASELINE = 0.0284   # 7-day unprompted purchase rate after abandonment (119 / 4,189)
DAILY    = 45.5     # eligible abandoners per day (4,189 / 92 days)
ALPHA    = 0.05
POWER    = 0.80

print(f"baseline = {BASELINE:.4f}, eligible/day = {DAILY}, alpha = {ALPHA}, power = {POWER}\n")
print(f"{'MDE':>8} {'target':>8} {'n/arm':>8} {'days':>6} {'weeks':>6}")

for mde in [0.010, 0.015, 0.020, 0.030]:
    effect = proportion_effectsize(BASELINE, BASELINE + mde)
    n = NormalIndPower().solve_power(
        effect_size=effect, alpha=ALPHA, power=POWER, alternative="two-sided"
    )
    days = 2 * n / DAILY
    print(f"+{mde*100:.1f}pp {BASELINE+mde:>8.4f} {math.ceil(n):>8,} {days:>6.0f} {days/7:>6.1f}")
```
